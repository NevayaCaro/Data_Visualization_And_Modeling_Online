```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv('Employee_productivity.csv')

df.head()
df.info()
```

```python
numeric_cols = df.select_dtypes(include='number').columns
categorical_cols = df.select_dtypes(exclude='number').columns

numeric_cols, categorical_cols
```

```python
prod_col = [c for c in numeric_cols if 'product' in c.lower()][0]
prod_col
```

```python
years_col = [c for c in numeric_cols if 'year' in c.lower()][0]
hours_col = [c for c in numeric_cols if 'hour' in c.lower()][0]
focus_col = [c for c in numeric_cols if 'focus' in c.lower()][0]

years_col, hours_col, focus_col
```

```python
sns.regplot(x=years_col, y=prod_col, data=df)
plt.title('Years at Company vs Productivity')
plt.show()

sns.regplot(x=hours_col, y=prod_col, data=df)
plt.title('Hours Worked vs Productivity')
plt.show()

sns.regplot(x=focus_col, y=prod_col, data=df)
plt.title('Focus vs Productivity')
plt.show()
```

```python
def calc_r2(x, y):
    model = LinearRegression()
    model.fit(x.values.reshape(-1,1), y)
    preds = model.predict(x.values.reshape(-1,1))
    return r2_score(y, preds)
```

```python
r2_years = calc_r2(df[years_col], df[prod_col])
r2_hours = calc_r2(df[hours_col], df[prod_col])
r2_focus = calc_r2(df[focus_col], df[prod_col])

r2_years, r2_hours, r2_focus
```

```python
for col in numeric_cols:
    if col != prod_col:
        sns.regplot(x=col, y=prod_col, data=df)
        plt.title(f'{col} vs Productivity')
        plt.show()
```

```python
for col in categorical_cols:
    sns.boxplot(x=col, y=prod_col, data=df)
    plt.title(f'{col} vs Productivity')
    plt.xticks(rotation=45)
    plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
X = df[[years_col, hours_col, focus_col]]
y = df[prod_col]

model = LinearRegression()
model.fit(X, y)

preds = model.predict(X)
r2_model = r2_score(y, preds)

r2_model
```

```python
Focus has the strongest relationship with 
productivity, as shown by both the visualization 
and the highest R-squared value. While working more
hours can help up to a point, the results suggest 
there are diminishing returns, meaning longer hours
do not always lead to higher productivity. Years at 
the company shows a much weaker relationship, 
indicating that experience alone does not guarantee 
better performance. The data also suggests that
frequent meetings and distracting work environments
can reduce productivity. To improve overall 
productivity, the company should limit unnecessary 
meetings, encourage uninterrupted focus time, and 
provide more flexibility through remote or quieter
work options.
```
