```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('earthquakes.csv')
df.head()
```

```python
This dataset comes from the United States
Geological Survey (USGS) and includes information 
on earthquakes from around the world. Each row
represents a single earthquake and records where
and when it occurred using latitude, longitude,
    and a timestamp. The dataset also includes 
details such as the earthquake’s magnitude, depth, 
and a brief description of its location.
```

```python
df['time'] = pd.to_datetime(df['time'])
df['year'] = df['time'].dt.year
```

```python
df.groupby('year').size().plot(kind='line')
plt.title('Number of Earthquakes Over Time')
plt.xlabel('Year')
plt.ylabel('Number of Earthquakes')
plt.show()

```

```python
This visualization shows how the number of 
recorded earthquakes changes over time. Although 
earthquake activity itself does not appear to 
increase dramatically, the rise in reported events 
in recent years is likely due to improvements in 
monitoring and detection technology rather than a 
true increase in earthquakes.
```

```python
plt.scatter(df['longitude'], df['latitude'], alpha=0.3, s=10)
plt.title('Global Distribution of Earthquakes')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

```python
This map shows where earthquakes occur around 
the world. Most earthquakes are clustered along
tectonic plate boundaries, especially around
the Pacific Ring of Fire, rather than being 
evenly spread across the globe. This pattern
highlights the strong connection between 
earthquake activity and plate movement.
```

```python
sns.scatterplot(x='year', y='mag', data=df, alpha=0.4)
plt.title('Earthquake Magnitude Over Time')
plt.xlabel('Year')
plt.ylabel('Magnitude')
plt.show()
```

```python
This visualization shows how earthquake magnitudes 
change over time. Larger earthquakes appear 
occasionally throughout the timeline, but there 
is no consistent increase or decrease in magnitude 
overall. This suggests that while earthquakes happen
at different times and intensities, their overall 
strength has remained relatively stable over the 
years.
```
