# Lab Instructions

You are part of a data team for a financial institution that uses an AI model to determine who will and will not be granted a loan.  Recently, there have been complaints from customers that the model unfairly discriminates based on the applicant's age and sex. A court has ordered that the financial institution must provide information about how the model works.  

Data from recent loan applications is given in `loan_data.csv`.  The data dictionary is below:

| Feature              | Description | Data Type |
| :---------------- | :------: | ----: |
|person_age|Age of the person|Float
|person_gender|Gender of the person|Categorical
|person_education|Highest education level|Categorical
|person_income|Annual income|Float
|person_emp_exp|Years of employment experience|Integer
|person_home_ownership|Home ownership status (e.g., rent, own, mortgage)|Categorical
|loan_amnt|Loan amount requested|Float
|loan_intent|Purpose of the loan|Categorical
|loan_int_rate|Loan interest rate|Float
|loan_percent_income|Loan amount as a percentage of annual income|Float
|cb_person_cred_hist_length|Length of credit history in years|Float
|credit_score|Credit score of the person|Integer
|previous_loan_defaults_on_file|Indicator of previous loan defaults|Categorical
|loan_status (target variable)|Loan approval status: 1 = approved; 0 = rejected|Integer

Visualize the relationship between age, sex and 5 other features with `loan_status`.  Which features seem to drive the model's decision to grant a loan?  Which seem unrelated?  Should the financial institution be concerned about potential legal trouble?  Make sure to justify your answers with specific references to your visualizations.



```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('loan_data.csv')
df.head()
```

```python
This dataset contains information about recent loan applications. The target variable is `loan_status` (1 = approved, 0 = rejected). 
We will visualize relationships between `loan_status` and applicant features such as age, gender, income, and other key attributes.
```

```python
Age vs Loan Status 
sns.boxplot(x='loan_status', y='person_age', data=df)
plt.title('Person Age vs Loan Status')
plt.show()

Gender vs Loan Status
sns.countplot(x='person_gender', hue='loan_status', data=df)
plt.title('Person Gender vs Loan Status')
plt.show()

Income vs Loan Status
sns.boxplot(x='loan_status', y='person_income', data=df)
plt.title('Person Income vs Loan Status')
plt.show()

Employment Experience vs Loan Status
sns.boxplot(x='loan_status', y='person_emp_exp', data=df)
plt.title('Employment Experience vs Loan Status')
plt.show()

Loan Amount vs Loan Status
sns.boxplot(x='loan_status', y='loan_amnt', data=df)
plt.title('Loan Amount vs Loan Status')
plt.show()

Credit Score vs Loan Status
sns.boxplot(x='loan_status', y='credit_score', data=df)
plt.title('Credit Score vs Loan Status')
plt.show()

Credit History Length vs Loan Status
sns.boxplot(x='loan_status', y='cb_person_cred_hist_length', data=df)
plt.title('Credit History Length vs Loan Status')
plt.show()
```

```python
- **Age:** Older applicants seem slightly more likely to be approved, but there is no extreme bias. 
- **Gender:** There is a noticeable difference in approval rates between genders, which could indicate potential discrimination.
- **Income:** Higher income correlates strongly with loan approval.
- **Employment Experience:** Longer employment history correlates with higher approval probability.
- **Loan Amount:** Larger loan requests are less likely to be approved.
- **Credit Score:** Higher credit scores strongly predict approval.
- **Credit History Length:** Longer credit history correlates with approval.
```

```python
Based on visualizations:
- Strongly related features: `person_income`, `credit_score`, `loan_amnt`, `cb_person_cred_hist_length`
- Moderately related: `person_age`, `person_emp_exp`
- Potentially unrelated: `person_gender` (though could be legally sensitive)
```

```python
The model shows differences in approval rates based on gender. This could be considered **discrimination against a protected class**, which is a legal concern. Age also shows minor trends, which might require review to ensure compliance with anti-discrimination laws. 

The financial institution should review the model for bias and implement fairness checks to avoid potential legal issues.
```

```python
numeric_features = ['person_age', 'person_income', 'person_emp_exp', 'loan_amnt', 
                    'loan_int_rate', 'loan_percent_income', 'cb_person_cred_hist_length', 
                    'credit_score']

corr = df[numeric_features + ['loan_status']].corr()
plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation of Numeric Features with Loan Status')
plt.show()
```

```python
This heatmap confirms that `credit_score` and `person_income` have the strongest positive correlation with loan approval, while `loan_amnt` is negatively correlated.
```
