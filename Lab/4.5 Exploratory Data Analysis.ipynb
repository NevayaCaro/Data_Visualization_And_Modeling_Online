```python
# Titanic Dataset Analysis Lab

In this lab, I selected the Titanic passenger 
dataset to explore relationships between different
features and a target variable. The goal is to 
determine which features are most related to
survival and which have little influence.
```

```python
## Dataset Description

**Dataset:** Titanic Passenger Dataset  
**Number of rows:** 891  
**Number of features:** More than 5
    (only 5 selected for analysis)

The dataset contains information about
passengers aboard the Titanic, including 
demographic information, ticket details,
and survival outcomes.

    
```

```python
## Target Feature

**Target Feature:** `Survived`

**Values:**
- 0 = Did not survive
- 1 = Survived

This is a binary target variable representing
whether a passenger survived the Titanic disaster.
```

```python
## Selected Features

The following five features were selected for analysis:

1. **Pclass**
   - Values: 1 (First), 2 (Second), 3 (Third)

2. **Sex**
   - Values: male, female

3. **Age**
   - Values: Continuous numerical values 
(approximately 0–80)

4. **Fare**
   - Values: Continuous numerical values 
(approximately 0–512)

5. **Embarked**
   - Values: C (Cherbourg), Q (Queenstown), S 
(Southampton)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset
titanic = sns.load_dataset("titanic")

# Display first few rows
titanic.head()
```

```python
## Visualization 1: Survival vs Sex

This bar chart shows the survival rate grouped by passenger gender.
```

```python
sns.barplot(x="sex", y="survived", data=titanic)
plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")
plt.show()
```

```python
**Interpretation:**  
Female passengers had a much higher survival 
rate than male passengers. This suggests that 
gender was one of the strongest factors influencing 
survival.
```

```python
## Visualization 2: Survival vs Passenger Class

This bar chart shows survival rates across 
passenger classes.
```

```python
**Interpretation:**  
Passengers in first class had the highest 
survival rate, while third class passengers 
had the lowest. Passenger class appears to
strongly influence survival.
```

```python
## Visualization 3: Survival vs Age

This histogram compares the age distribution 
of survivors and non-survivors.
```

```python
sns.histplot(data=titanic, x="age", hue="survived",
bins=30)
plt.title("Age Distribution by Survival")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
**Interpretation:**  
Younger passengers, especially children, appear
more likely to survive. However, age alone is not 
as strong a predictor as gender or class.
```

```python
## Visualization 4: Survival vs Fare

This boxplot compares ticket prices paid by 
survivors and non-survivors.
```

```python
sns.boxplot(x="survived", y="fare", data=titanic)
plt.title("Fare Distribution by Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Fare")
plt.show()
```

```python
**Interpretation:**  
Survivors generally paid higher fares, 
suggesting that wealthier passengers had a
higher chance of survival.
```

```python
## Visualization 5: Survival vs Embarked

This bar chart shows survival rates based on
embarkation port.
```

```python
sns.barplot(x="embarked", y="survived", data=titanic)
plt.title("Survival Rate by Embarkation Port")
plt.xlabel("Embarked")
plt.ylabel("Survival Rate")
plt.show()
```

```python
**Interpretation:**  
Passengers who embarked at Cherbourg (C) had 
higher survival rates. This is likely due to a
higher number of first-class passengers boarding 
there.
```

```python
## Conclusion

Based on the visualizations, survival on the 
Titanic was strongly influenced by gender, passenger
class, and fare. Sex appears to be the most influential feature, followed by passenger class and ticket price. Age and embarkation port have some influence but are less predictive overall. The data suggests that social factors played a major role in survival outcomes.
```
